## Section 2: Grounded RAG Pipeline
Semantic section chunking, asymmetric embedding generation, cosine similarity retrieval, and grounded generation.

```python
# Step 2: Semantic Chunking & Vector Embeddings
import numpy as np

def chunk_document(source_name: str, text: str):
    sections = text.split("\n---")
    results = []
    for idx, sec in enumerate(sections):
        cleaned = sec.strip()
        if len(cleaned) > 20:
            results.append({"source": source_name, "chunk_id": f"{source_name}#sec{idx+1}", "text": cleaned})
    return results

all_chunks = []
for blob in bucket.list_blobs():
    if blob.name.endswith(".txt"):
        all_chunks.extend(chunk_document(blob.name, blob.download_as_text()))

doc_inputs = [TextEmbeddingInput(text=c["text"], task_type="RETRIEVAL_DOCUMENT") for c in all_chunks]
chunk_embeddings = embedding_model.get_embeddings(doc_inputs)
for i, emb in enumerate(chunk_embeddings):
    all_chunks[i]["vector"] = np.array(emb.values)

def retrieve_context(query: str, top_k: int = 2):
    query_input = TextEmbeddingInput(text=query, task_type="RETRIEVAL_QUERY")
    q_vec = np.array(embedding_model.get_embeddings([query_input])[0].values)
    scored = []
    for chunk in all_chunks:
        dot = np.dot(q_vec, chunk["vector"])
        norm = np.linalg.norm(q_vec) * np.linalg.norm(chunk["vector"])
        sim = float(dot / norm) if norm > 0 else 0.0
        scored.append((sim, chunk))
    scored.sort(key=lambda x: x[0], reverse=True)
    return scored[:top_k]

print(f"Indexed {len(all_chunks)} semantic chunks in memory.")
```

```python
# Step 2.5: Evaluate Grounded RAG Generation
def ask_cloudcorp_assistant(query: str, top_k: int = 2):
    top_hits = retrieve_context(query, top_k=top_k)
    context_str = "\n\n".join([f"[{h[1]['chunk_id']}]:\n{h[1]['text']}" for h in top_hits])
    prompt = f"""You are CloudCorp's AskHR Assistant. Answer the employee's question strictly using ONLY the policy context below.
Always cite the chunk_id. If the answer cannot be found in the context, state that you do not know.

Context:
{context_str}

Question: {query}
Answer:"""
    return flash_model.generate_content(prompt).text.strip()

print(ask_cloudcorp_assistant("How many days of PTO can I carry over, and when do they expire?"))
```

## Section 3: Extending RAG to an Agentic AI Workflow
Autonomous tool declaration and multi-step reasoning loop across the 3 core business scenarios.

```python
# Step 3: Implement IT Tool & FunctionDeclaration
TICKETS_DB = {
    "INC-402": {"ticket_id": "INC-402", "status": "In Progress", "assigned_to": "Hardware Desk", "eta": "Tomorrow 2:00 PM"},
    "INC-105": {"ticket_id": "INC-105", "status": "Resolved", "assigned_to": "Network Ops", "eta": "Completed"}
}
TICKET_COUNTER = 501

def create_it_ticket(issue_summary: str, category: str = "Hardware Replacement") -> dict:
    global TICKET_COUNTER
    ticket_id = f"INC-{TICKET_COUNTER}"
    TICKET_COUNTER += 1
    record = {
        "ticket_id": ticket_id,
        "status": "Created (Approved for Loaner)",
        "category": category,
        "summary": issue_summary,
        "assigned_to": "Hardware Fulfillment Desk (Floor 2)",
        "priority": "High (P2)",
        "eta": "Within 24 hours",
        "instructions": "Present ticket ID at Floor 2 Tech Depot for emergency loaner laptop."
    }
    TICKETS_DB[ticket_id] = record
    return record

def lookup_it_ticket(ticket_id: str) -> dict:
    tid = ticket_id.upper()
    return TICKETS_DB.get(tid, {"ticket_id": tid, "status": "Not Found"})

create_ticket_func = FunctionDeclaration(
    name="create_it_ticket",
    description="Creates a new IT incident ticket when an employee reports damaged or malfunctioning hardware.",
    parameters={
        "type": "object",
        "properties": {
            "issue_summary": {"type": "string", "description": "Summary of hardware malfunction."},
            "category": {"type": "string", "description": "Category of ticket."}
        },
        "required": ["issue_summary"]
    }
)

lookup_ticket_func = FunctionDeclaration(
    name="lookup_it_ticket",
    description="Checks live progress for an IT incident ticket using its ticket ID.",
    parameters={
        "type": "object",
        "properties": {
            "ticket_id": {"type": "string", "description": "Ticket ID (e.g., INC-402, INC-501)."}
        },
        "required": ["ticket_id"]
    }
)

agent_tools = Tool(function_declarations=[create_ticket_func, lookup_ticket_func])
agent_llm = vertexai.init(project=PROJECT_ID, location="global", credentials=credentials)
agent_llm = GenerativeModel(
    "gemini-3.5-flash-lite",
    tools=[agent_tools],
    system_instruction=(
        "You are CloudCorp's AskHR & IT Agent. "
        "When an employee reports broken hardware, explain replacement policy and call create_it_ticket. "
        "When an employee asks about a ticket ID, call lookup_it_ticket. "
        "For general policy questions, answer directly from policy context without calling tools."
    )
)
print("AskHR Agent initialized with external tools.")
```

```python
# Step 3.4: Execute Agent Across 3 Scenarios
def run_askhr_agent(user_query: str):
    print(f"\n================ USER QUERY: '{user_query}' ================")
    rag_hits = retrieve_context(user_query, top_k=1)
    context_text = rag_hits[0][1]["text"] if rag_hits else "No specific policy chunk retrieved."
    
    chat = agent_llm.start_chat()
    prompt = f"Policy Context:\n{context_text}\n\nUser Request: {user_query}"
    response = chat.send_message(prompt)
    
    part = response.candidates[0].content.parts[0]
    if hasattr(part, "function_call") and part.function_call.name:
        fc = part.function_call
        fn_name = fc.name
        fn_args = dict(fc.args)
        print(f"⚡ [AGENT TOOL CALL]: {fn_name}({fn_args})")
        
        if fn_name == "create_it_ticket":
            tool_res = create_it_ticket(**fn_args)
        elif fn_name == "lookup_it_ticket":
            tool_res = lookup_it_ticket(**fn_args)
        else:
            tool_res = {"error": "Unknown tool"}
            
        print(f"📦 [TOOL RESULT]: {tool_res}")
        response_part = Part.from_function_response(name=fn_name, response={"content": tool_res})
        final_resp = chat.send_message(response_part)
        print(f"🤖 [FINAL SYNTHESIZED RESPONSE]:\n{final_resp.text.strip()}")
    else:
        print(f"ℹ️ [PURE RAG PATH (No Tools Needed)]")
        print(f"🤖 [RESPONSE]:\n{response.text.strip()}")

run_askhr_agent("What is remote work policy?")
run_askhr_agent("My laptop screen is broken and flickering, what should I do?")
run_askhr_agent("What is the status of my ticket INC-501?")
```

## Section 4: Production Architecture & Cost Controls
Matryoshka Representation Learning (MRL) dimension reduction and cost auditing.

```python
import numpy as np
# Step 4: Matryoshka Representation Learning (768 -> 256) & Cost Auditor
full_emb = embedding_model.get_embeddings(["CloudCorp hybrid work stipend reimbursement"])[0]
full_vec = np.array(full_emb.values)
mrl_256_vec = full_vec[:256]
mrl_256_vec = mrl_256_vec / np.linalg.norm(mrl_256_vec)

print(f"Full Vector Dimensions   : {len(full_vec)} floats ({len(full_vec) * 4} bytes)")
print(f"MRL Vector Dimensions    : {len(mrl_256_vec)} floats ({len(mrl_256_vec) * 4} bytes)")
print(f"Memory / Index Savings   : {(1 - len(mrl_256_vec) / len(full_vec)) * 100:.1f}%")

# Lab Cost Calculation
total_cost = (25 * 600 / 1000 * 0.000025) + (30 * 800 / 1_000_000 * 0.075) + (30 * 150 / 1_000_000 * 0.30) + 0.0005
print(f"\nTotal Projected Lab Cost: ${total_cost:.4f} (Budget Cap: $5.00)")
```
